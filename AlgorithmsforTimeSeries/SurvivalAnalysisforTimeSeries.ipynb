Survival Analysis

There are two types of learning from data based methods 
1. Parameteric Method
2. Non-parameteric Method

Parameteric survival analysis models like Weibull, exponential, log-linear, and many others.

Non-parametric survival analysis models like Kaplan-Meier and Nelson-Aalen and many more.

Kaplan-Meier model 
- measures a binary variable using time.
- assumptions about the structure of data.
- assumptions about linearity and normality of data.
- It captures the actual entry and exit date.
- Note : Duration more importent than the time.
