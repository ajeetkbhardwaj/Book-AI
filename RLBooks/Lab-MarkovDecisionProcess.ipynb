### 1. Markov Chain

```python
import numpy as np
# Define the states and transition matrix of markov chain
states = ["Sunny", "Rainy"]
transition_matrix = np.array([[0.8, 0.2], [0.4, 0.6]])
# Initial state
initial_state = 1 # Sunny
# Simulate the Markov chain for 10 steps
current_state = initial_state
for _ in range(10):
   print(states[current_state], end=" -> ")
   current_state = np.random.choice([0, 1], p=transition_matrix[current_state])
print("End")
```

```python
import numpy as np

P = np.array([[0.3, 0.7],
              [0.2, 0.8]])

S = np.array([0.5, 0.5])

for i in range(10):
    S = np.dot(S, P)
    print(f"Iter {i}: S = {S}")

print("Final steady state:", S)
```

Why above markov chain is converges ? Because markov chain is irreducible and aperiodic.
